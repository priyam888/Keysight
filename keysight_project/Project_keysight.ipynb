```python
import sys
import os

TASKS_FILE = "tasks.txt"
```

# Load tasks from the file

```python
def load_tasks():
    if not os.path.exists(TASKS_FILE):
        return []
    with open(TASKS_FILE, 'r') as f:
        return [task.strip() for task in f.readlines()]
```

# Save tasks to the file

```python
def save_tasks(tasks):
    with open(TASKS_FILE, 'w') as f:
        for task in tasks:
            f.write(task + '\n')
```

# Add a new task

```python
def add_task(task_description):
    tasks = load_tasks()
    tasks.append(task_description)
    save_tasks(tasks)
    print(f"Added task: {task_description}")
```

# List all pending tasks

```python
def list_tasks():
    tasks = load_tasks()
    if not tasks:
        print("No tasks found.")
        return
    print("Pending tasks:")
    for i, task in enumerate(tasks, start=1):
        print(f"{i}. {task}")
```

# Mark a task as complete

```python
def complete_task(task_number):
    tasks = load_tasks()
    try:
        task = tasks.pop(task_number - 1)
        save_tasks(tasks)
        print(f"Completed task: {task}")
    except IndexError:
        print("Invalid task number.")
```

# Delete a task

```python
def delete_task(task_number):
    tasks = load_tasks()
    try:
        task = tasks.pop(task_number - 1)
        save_tasks(tasks)
        print(f"Deleted task: {task}")
    except IndexError:
        print("Invalid task number.")
```

# Main function to handle command-line arguments

```python
def main():
    if len(sys.argv) < 2:
        print("Usage: python task_manager.py [add/list/complete/delete] [task description/number]")
        return

    command = sys.argv[1]

    if command == "add":
        if len(sys.argv) < 3:
            print("Please provide a task description.")
        else:
            add_task(' '.join(sys.argv[2:]))
    elif command == "list":
        list_tasks()
    elif command == "complete":
        if len(sys.argv) < 3:
            print("Please provide the task number to complete.")
        else:
            try:
                complete_task(int(sys.argv[2]))
            except ValueError:
                print("Task number should be a valid integer.")
    elif command == "delete":
        if len(sys.argv) < 3:
            print("Please provide the task number to delete.")
        else:
            try:
                delete_task(int(sys.argv[2]))
            except ValueError:
                print("Task number should be a valid integer.")
    else:
        print(f"Unknown command: {command}")
```

```python
if __name__ == "__main__":
    main()
```
